```python
# Program Kasir Sederhana Kedai Miftaah

def kasir():
    nama = "Hidayatil"
    harga_makanan = 15000
    jumlah_makanan = 3

    subtotal = harga_makanan * jumlah_makanan
    diskon = subtotal * 0.10
    total_bayar = subtotal - diskon

    print("===== KEDAI MIFTAAH =====")
    print("Nama pelanggan :", nama)
    print("Harga makanan  :", harga_makanan)
    print("Jumlah pesanan :", jumlah_makanan)
    print("Subtotal       :", subtotal)
    print("Diskon 10%     :", diskon)
    print("Total bayar    :", total_bayar)
    print("=========================")
    print("Terima kasih sudah berbelanja!")

kasir()
```
