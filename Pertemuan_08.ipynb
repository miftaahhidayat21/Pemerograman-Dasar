```python
#6F
skor = [80, 65, 90, 75, 70]

total = 0
jumlah_lulus = 0

for nilai in skor:
    total = total + nilai
    if nilai >= 75:
        jumlah_lulus = jumlah_lulus + 1
    print("total:", total)

rata_rata = total / len(skor)

print("Total      :", total)
print("Rata-rata  :", rata_rata)
print("Skor >= 75:", jumlah_lulus, "dari", len(skor))
```

```python
#6G
suhu = [28, 31, 30, 33, 29, 27, 32]

# Perubahan harian: mulai dari indeks 1
for i in range(1, len(suhu)):
    perubahan = suhu[i] - suhu[i - 1]
    print(f"Hari {i + 1}: {suhu[i]} °C, perubahan {perubahan:+d}")

# Rata-rata bergerak 5 hari
rata5 = []
for i in range(4, len(suhu)):
    jendela = suhu[i - 4 : i + 1]
    rata5.append(round(sum(jendela) / 5, 1))

print("Rata-rata 5 hari:", rata5)
print("len(suhu) =", len(suhu), ", nilai =", len(rata5), "rata-rata")
```

```python
# LATIHAN PRAKTIKUM
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062
]

# Total dan rata-rata energi
total = sum(energi)
rata_rata = total / len(energi)

print("Total energi bulanan:", total, "MWh")
print("Rata-rata energi:", round(rata_rata, 2), "MWh")

# 3 hari tertinggi
tertinggi = sorted(energi, reverse=True)[:3]
print("3 energi tertinggi:", tertinggi)

# 3 hari terendah (hari dan nilainya)
terendah = sorted(enumerate(energi, start=1), key=lambda x: x[1])[:3]
print("3 energi terendah (hari, nilai):", terendah)

# Jumlah hari di bawah target
target = int(input("Masukkan target energi harian (MWh): "))
jumlah_di_bawah = 0

for nilai in energi:
    if nilai < target:
        jumlah_di_bawah += 1

print("Jumlah hari di bawah target:", jumlah_di_bawah)

# Rata-rata bergerak 7 hari yang di bawah 97% rata-rata bulanan
batas = rata_rata * 0.97
rata7 = []

for i in range(6, len(energi)):
    jendela = energi[i - 6:i + 1]
    rata = sum(jendela) / 7
    rata7.append(rata)

hari_di_bawah_97 = []

for i in range(len(rata7)):
    if rata7[i] < batas:
        hari_di_bawah_97.append(i + 7)

print("Batas 97% rata-rata bulanan:", round(batas, 2), "MWh")
print("Hari rata-rata 7 hari di bawah batas:", hari_di_bawah_97)

# Total energi per minggu (7 hari), dan sisa hari
minggu1 = sum(energi[0:7])
minggu2 = sum(energi[7:14])
minggu3 = sum(energi[14:21])
minggu4 = sum(energi[21:28])
sisa = sum(energi[28:30])

print("Energi minggu 1:", minggu1, "MWh")
print("Energi minggu 2:", minggu2, "MWh")
print("Energi minggu 3:", minggu3, "MWh")
print("Energi minggu 4:", minggu4, "MWh")
print("Energi sisa hari:", sisa, "MWh")
```
