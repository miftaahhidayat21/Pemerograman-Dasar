```python
#SOAL 1
jumlah_buku = int(input("Jumlah buku: "))
harga_buku = int(input("Harga per buku: "))
diskon_persen = int(input("Diskon (%): "))
uang_budi = int(input("Uang Budi: "))
jumlah_orang = int(input("Jumlah orang: "))

total = jumlah_buku * harga_buku
diskon = total * diskon_persen // 100
bayar = total - diskon
kembalian = uang_budi - bayar
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang
cukup = uang_budi >= bayar

print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", cukup)
```

```python
#SOAL 2
jumlah_panel = int(input("Jumlah panel: "))
daya_panel = float(input("Daya per panel (Wp): "))
jam_matahari = float(input("Penyinaran efektif (jam): "))
efisiensi = float(input("Efisiensi sistem: "))
kebutuhan = float(input("Kebutuhan listrik (kWh/hari): "))
tarif = float(input("Tarif listrik (Rp/kWh): "))

energi_harian = jumlah_panel * daya_panel / 1000 * jam_matahari * efisiensi
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif
kekurangan = kebutuhan - energi_harian

if kekurangan < 0:
    kekurangan = 0

mencukupi = energi_harian >= kebutuhan

print("Energi harian (kWh) :", round(energi_harian, 2))
print("Energi bulanan (kWh):", round(energi_bulanan, 2))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 2))
print("Kekurangan (kWh)    :", round(kekurangan, 2))
print("Mencukupi?          :", mencukupi)
```

```python
#SOAL 3
daya_generator = float(input("Daya generator (kW): "))
jam_operasi = int(input("Jam operasi per hari: "))
konsumsi = float(input("Konsumsi solar (liter/kWh): "))
harga_solar = float(input("Harga solar per liter: "))
kapasitas_tangki = int(input("Kapasitas tangki (liter): "))
arus = float(input("Arus kabel (A): "))
hambatan = float(input("Hambatan kabel (ohm): "))

energi = daya_generator * jam_operasi
solar = energi * konsumsi
biaya = solar * harga_solar
hari_penuh = kapasitas_tangki // solar
sisa_solar = kapasitas_tangki - (hari_penuh * solar)
rugi_kabel = arus ** 2 * hambatan
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("Energi (kWh)     :", round(energi, 2))
print("Solar (liter)    :", round(solar, 2))
print("Biaya (Rp)       :", round(biaya, 2))
print("Tangki cukup     :", float(hari_penuh), "hari")
print("Sisa solar (L)   :", round(sisa_solar, 2))
print("Rugi kabel (W)   :", round(rugi_kabel, 2))
print("Rugi tinggi?     :", rugi_tinggi)
```

```python
#LATIHAN 1
# Memasukkan nilai dengan fungsi int() agar tipe datanya integer
jumlah_gorengan = int(input("Masukkan jumlah gorengan: "))
harga_gorengan = int(input("Masukkan harga satuan gorengan: "))
jumlah_esteh = int(input("Masukkan jumlah es teh: "))
harga_esteh = int(input("Masukkan harga satuan es teh: "))
uang = int(input("Masukkan jumlah uang yang dibayarkan: "))

# Hitung total harga masing-masing item
total_gorengan = jumlah_gorengan * harga_gorengan
total_esteh = jumlah_esteh * harga_esteh

# Menghitung Total Belanja
total_belanja = total_gorengan + total_esteh

# Menghitung Uang Kembalian
kembalian = uang - total_belanja

# Mengecek apakah uang cukup (True/False)
apakah_cukup = uang >= total_belanja

# Menampilkan hasil
print("\n--- Hasil Perhitungan ---")
print("Total Belanja:", total_belanja)
print("Uang Kembalian:", kembalian)
print("Apakah Uang Cukup?:", apakah_cukup)
```

```python
#LATIHAN 2
# Durasi Film Maraton

total_menit = 275

# Menghitung jam penuh dan menit sisa
jam_total = total_menit // 60
menit_sisa = total_menit % 60

# Total durasi dalam jam desimal
jam_desimal = total_menit / 60

# Mengecek apakah durasi lebih dari 3 jam dan kurang dari 5 jam
cek_durasi = (jam_desimal > 3) and (jam_desimal < 5)

print(f"{jam_total} jam {menit_sisa} menit")
print("Total durasi dalam jam desimal:", jam_desimal)
print("Apakah durasi lebih dari 3 jam dan kurang dari 5 jam?", cek_durasi)
```
