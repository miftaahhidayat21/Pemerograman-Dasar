```python
# Nama:Miftaah Hidayatil Hadi
# Kelas:Teknologi Rekayasa Pembangkit Energi Pagi B
# NIM:4232501059
# Tanggal:10 November 2026
# MataKuliah:Pemerograman Dasar
# Soal 2:Energi Harian dan Capacity Factor PLTD
# SOAL GENAP NIM 59

# Simpan nilai ke dalam variabel
daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# hitung energi harian
energi = daya_kw * jam

# hitung capacity factor
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

# Tampilkan hasilnya
print(f"Energi harian = {energi} kWh")
print(f"Capacity factor = {capacity_factor:.2f}%")
```

```python
# Nama:Miftaah Hidayatil Hadi
# Kelas:Teknologi Rekayasa Pembangkit Energi Pagi B
# NIM:4232501059
# Tanggal:10 November 2026
# MataKuliah:Pemerograman Dasar
# Soal 6:Status Operasi Boiler
# SOAL GENAP NIM 59

# Masukkan data suhu dan tekanan
suhu = float(input("Masukkan suhu uapnya berapa (°C): "))
tekanan = float(input("Masukkan tekanannya berapa (bar): "))

# Tentukan status operasi boiler
if suhu >= 550 or tekanan >= 14:
    print("STATUSNYA: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUSNYA: WARNING")
else:
    print("STATUSNYA: NORMAL")
```

```python
# Nama:Miftaah Hidayatil Hadi
# Kelas:Teknologi Rekayasa Pembangkit Energi Pagi B
# NIM:4232501059
# Tanggal:10 November 2026
# MataKuliah:Pemerograman Dasar
# Soal 8:Pengosongan Tangki Bahan Bakar
# SOAL GENAP NIM 59

# Tentukan volume awal dari tangki
volume = 100

# Perulangan selama volume masih lebih dari 0
while volume > 0:
    volume -= 20
    print(f"Volume: {volume} liter")

    if volume == 20:
        print("WARNING: Hampir kosong!")

# Pesan setelah tangki kosong
print("TANGKI KOSONG - Pengosongan berhenti.")
```

```python
# Nama:Miftaah Hidayatil Hadi
# Kelas:Teknologi Rekayasa Pembangkit Energi Pagi B
# NIM:4232501059
# Tanggal:10 November 2026
# MataKuliah:Pemerograman Dasar
# Soal 10:Pembacaan Suhu Bearing dengan Sensor Rusak
# SOAL GENAP NIM 59

data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for i, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
        jumlah_valid += 1

        if suhu > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

print(f"Jumlah data valid yang berhasil diperiksa: {jumlah_valid}")


```
