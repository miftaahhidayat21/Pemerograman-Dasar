```python
# Nama    : Miftaah Hidayatil Hadi
# NIM     : 4232501059
# Tanggal : 27/09/2026
# Tujuan  : Menentukan apakah seseorang boleh mengemudi sendiri berdasarkan umur dan kepemilikan SIM.

umur = 19
punya_sim = True

boleh_mengemudi = (umur >= 17) and punya_sim

# Latihan operator precedence dengan tanda kurung
hasil_precedence = (umur >= 17) and punya_sim

print(f"Boleh mengemudi sendiri: {boleh_mengemudi}")
print(f"Hasil ekspresi dengan tanda kurung: {hasil_precedence}")
```

1. Operator Precedence

Operator precedence adalah aturan urutan pengerjaan operator dalam Python.
Perkalian (*) memiliki prioritas lebih tinggi daripada penjumlahan (+),
sehingga 10 + 5 * 2 dikerjakan sebagai 10 + (5 * 2) dan hasilnya adalah 20.
Jika ingin penjumlahan dilakukan terlebih dahulu, digunakan tanda kurung
(10 + 5) * 2, sehingga hasilnya 30.

Sumber: Python Documentation – Operator Precedence.

2. Operator Relasional dan Operator Logika

Operator relasional digunakan untuk membandingkan nilai dan menghasilkan True atau False, contohnya >, <, >=, dan ==. Sedangkan operator logika seperti and, or, dan not digunakan untuk menggabungkan beberapa kondisi. Contohnya, seseorang boleh mengemudi jika umur sudah 17 tahun dan memiliki SIM, yaitu (umur >= 17) and punya_sim.

3. Kesalahan Operator Precedence

Kesalahan operator precedence dapat terjadi ketika programmer lupa menggunakan tanda kurung pada kondisi yang menggunakan and dan or. Karena and memiliki prioritas lebih tinggi daripada or, hasil ekspresi dapat berbeda dari yang diinginkan. Contohnya, (username == "A" or username == "B") and msg == "!load" menggunakan tanda kurung agar kondisi username diperiksa terlebih dahulu. Dengan begitu, urutan operasi menjadi jelas dan hasil program sesuai dengan tujuan.
